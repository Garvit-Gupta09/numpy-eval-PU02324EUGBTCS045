```python
import numpy as np

np.random.seed(42)

A = np.random.randint(1, 11, size=(4, 3))
B = np.random.randint(1, 11, size=(3, 5))

print("Matrix A:")
print(A)
print("Shape of A:", A.shape)

print("\nMatrix B:")
print(B)
print("Shape of B:", B.shape)

assert A.shape[1] == B.shape[0], "Inner dimensions do not match"
print("\nDimension validation successful.")

C = np.dot(A, B)

print("\nMatrix C = A · B:")
print(C)
print("Shape of C:", C.shape)

assert C.shape == (4, 5)

total_sum = np.sum(C)
print("\nTotal sum of all elements:", total_sum)

row_sums = np.sum(C, axis=1)
print("\nRow-wise sums:")
print(row_sums)

column_sums = np.sum(C, axis=0)
print("\nColumn-wise sums:")
print(column_sums)
```
